Formatting with the .format() method

```python
print('This is a string {}'.format('INSERTED'))
```

```python
'The {} {} {}'.format('fox','quick','brown')
```

```python
# how to change order of format
'The {1} {2} {0}'.format('fox','quick','brown')
```

```python
'The {2} {2} {2}'.format('fox','quick','brown')
```

```python
'The {q} {b} {f}'.format(f='fox',q='quick',b='brown')
```

Float formatting "{value:width.precision f}"

```python
result = 10000/700
```

```python
result
```

```python
'This is a formated result {r:1.5f}'.format(r=result)
```

```python
f'This is a formated result {result:1.3f}'
```
