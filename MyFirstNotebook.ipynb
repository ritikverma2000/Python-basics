Reading , Writing, Appending Modes

mode = 'r' 
mode ='w'
mode='a'
mode='r+' is reading and writing
mode 'w+' is writing and reading

```python
%%writefile my_new_file.txt
ONE ON FIRST
TWO ON SECOND
THREE ON THIRD
```

```python
with open('my_new_file.txt',mode='r') as f:
    print(f.read())
```

```python
with open('my_new_file.txt',mode='a') as f:
    f.write('\nFOUR ON FOURTH')
```

```python
with open('my_new_file.txt',mode='r') as f:
    print(f.read())
```

```python
with open('ritikRough.txt',mode='w') as f:
    f.write('Created this file')
```

```python
with open('ritikRough.txt',mode='r') as f:
    print(f.read())
```
